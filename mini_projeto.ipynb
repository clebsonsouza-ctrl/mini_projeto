importação

```python
#pip install pandas
import pandas as pd
```

```python
# Importando os dados
df = pd.read_csv("Base Varejo.csv", sep=";")

# Quantidade de registros e colunas
print("Total de registros:", df.shape[0])
print("Total de colunas:", df.shape[1])
print(30*'-')

# Tipos de dados
print("Tipos de dados:")
print(df.info())

# Visualizando os primeiros registros
display(df)
```

Tratamento de strings

```python
# Lista das colunas que possuem informações textuais
colunas_texto = ["CL_GENERO", "CL_SEG", "PR_CAT", "PR_NOME"]

# Percorre cada coluna da lista
for coluna in colunas_texto:

    # Aplica as transformações na coluna atual
    df[coluna] = (
        df[coluna]

        # Remove espaços no início e no final dos textos
        .str.strip()

        # Converte todos os caracteres para maiúsculas
        .str.upper()

        # Substitui vários espaços consecutivos por um único espaço
        .str.replace(r"\s+", " ", regex=True)
    )

# Exibe as 10 primeiras linhas das colunas tratadas
display(df[colunas_texto].head(10))
```

Tratamento de inteiros (Integer)

```python
# Colunas numéricas
colunas_numericas = [
    "CO_ID", "CL_ID", "CL_EC", "CL_FHL", "PR_ID"
]

# Garantindo o formato numérico
for coluna in colunas_numericas:
    df[coluna] = pd.to_numeric(df[coluna], errors="coerce")

# Conferindo os tipos
print(df[colunas_numericas].dtypes)
```

Tratamento de datas (Datetime)

```python
# Convertendo a coluna DATA
df["DATA"] = pd.to_datetime(
    df["DATA"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Verificando a transformação
print(df["DATA"].head())
print(df["DATA"].dtype)
```

Limpeza de Nulos e Duplicatas

```python
# Contando os valores nulos de cada coluna
print(df.isna().sum())

# Verificando valores representados por #N/D
print("Categorias sem informação:", (df["PR_CAT"] == "#N/D").sum())
print("Produtos sem informação:", (df["PR_NOME"] == "#N/D").sum())
```

Substituição de valores ausentes

```python
# Substituindo #N/D por valores nulos
df["PR_CAT"] = df["PR_CAT"].replace("#N/D", pd.NA)
df["PR_NOME"] = df["PR_NOME"].replace("#N/D", pd.NA)

# Tratando categorias sem informação
if df["PR_CAT"].isna().sum() > 0:

    # Preenche os valores ausentes
    df["PR_CAT"] = df["PR_CAT"].fillna("Sem Categoria")
else:
    print("Não existem categorias vazias.")

# Tratando nomes de produtos sem informação
if df["PR_NOME"].isna().sum() > 0:

    # Preenche os valores ausentes
    df["PR_NOME"] = df["PR_NOME"].fillna("Produto Desconhecido")
else:
    print("Não existem produtos vazios.")

# Conferindo os valores nulos após o tratamento
print(df.isna().sum())
```

Identificação de registros duplicados

```python
# Contando registros completamente duplicados
duplicatas = df.duplicated().sum()

# Exibindo a quantidade encontrada
print("Total de registros duplicados:", duplicatas)

# Visualizando alguns registros duplicados
display(df[df.duplicated()].head(10))
```

```python
# Criando uma cópia sem linhas completamente duplicadas
df_limpo = df.drop_duplicates().copy()

# Comparando o número de registros
print("Registros originais:", len(df))
print("Registros após limpeza:", len(df_limpo))

# Verificando se ainda existem duplicatas
print("Duplicatas restantes:", df_limpo.duplicated().sum())
```

Vamos verificar a quantidade de compras distintas

```python
# Contando os identificadores de compras diferentes
total_compras = df_limpo["CO_ID"].nunique()

# Verificando possíveis datas inválidas
datas_invalidas = df_limpo["DATA"].isna().sum()

# Exibindo os resultados
print("Compras distintas:", total_compras)
print("Datas inválidas:", datas_invalidas)
```

Estatística Descritiva

```python
# Criando um DataFrame com apenas um registro por cliente
df_clientes = df_limpo.drop_duplicates(subset="CL_ID").copy()

# Selecionando a coluna de número de filhos
filhos = df_clientes["CL_FHL"]

# Verificando a quantidade de clientes
print("Total de clientes:", len(df_clientes))
```

```python
# Contagem de clientes com informação
print("Contagem:", filhos.count())

# Medidas de tendência central
print("Média:", round(filhos.mean(), 2))
print("Mediana:", filhos.median())
print("Moda:", filhos.mode()[0])

# Medida de dispersão
print("Desvio padrão:", round(filhos.std(), 2))

# Valores extremos
print("Mínimo:", filhos.min())
print("Máximo:", filhos.max())

# Quartis da distribuição
print("1º Quartil:", filhos.quantile(0.25))
print("2º Quartil:", filhos.quantile(0.50))
print("3º Quartil:", filhos.quantile(0.75))
```

Distribuição do número de filhos

```python
# Contando clientes por número de filhos
distribuicao = filhos.value_counts().sort_index()

# Visualizando a distribuição
print(distribuicao)
```

Agrupamento por gênero dos clientes

```python
# Agrupa as compras pelo gênero dos clientes
# Conta apenas os identificadores de compra diferentes
compras_genero = df_limpo.groupby("CL_GENERO")["CO_ID"].nunique()

# Exibindo a quantidade de compras por gênero
print(compras_genero)

# F (Feminino): foram realizadas 9.615 compras distintas por clientes do gênero feminino.
# M (Masculino): foram realizadas 8.856 compras distintas por clientes do gênero masculino.
```

Agrupamento por categoria de produto

```python
# Agrupando os produtos por categoria
categorias = df_limpo.groupby("PR_CAT").agg({"CO_ID":"count"})

# Organizando da maior para a menor quantidade
categorias = categorias.sort_values(by= 'CO_ID', ascending= False)

# Exibindo o resultado
print(categorias)
```
